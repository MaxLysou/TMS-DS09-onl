```python
#класс для анализа датасета

class analysis_dataset:
  def __init__(self,sample):
    self.sample = sample

  #вывод n первых (order = head) и последних (order = tail) записей таблицы
  #аналоги head и tail из pandas
  def show(self, order: str, num: int) -> list:
    if num <= len(self.sample):

      if order == "head":
        return self.sample[0:num]
      elif order == "tail":
        return self.sample[len(self.sample)-num-1:len(self.sample)-1]
      else:
        print("Выбран неверный порядок! Укажите head или tail!")

    else:
      print("Количество выводимых символов превышает величину таблицы!")

  #вывод основных статистик descibe() для переменной
  def describe_of_variable(self, var1: str) -> list | None:
    if var1 not in self.sample.columns:
      print(f"Указанных переменных нет в анализируемом сэмпле!")
    else:
      return self.sample[var1].describe()

  #вывод статистик по двум переменным
  def table_with_stats(self,var1: str,var2: str) -> list | None:
    if var1 not in self.sample.columns or var2 not in self.sample.columns:
      print(f"Указанных переменных нет в анализируемом сэмпле!")
    else:
      return self.sample[[var1,var2]].groupby(var2).agg(["count","mean"])

  #вывод коэффициента корреляции между двумя переменными
  def correlation(self,var1: str,var2: str) -> list | None:
    if var1 not in self.sample.columns or var2 not in self.sample.columns:
      print(f"Указанных переменных нет в анализируемом сэмпле!")
    else:
      return self.sample[var1].corr(self.sample[var2])
```

```python
#создание объекта класса
sample1 = analysis_dataset(pd.read_csv('train.csv'))
#первые 10 значений
sample1.show("head",10)
```

```python
#последние 5 значений
sample1.show("tail",5)
```

```python
sample1.show("tall",5)
```

```python
sample1.show("tail",1000)
```

Анализ переменной Pclass

```python
sample1.describe_of_variable("Pclass")
```

```python
sample1.table_with_stats("Survived", "Pclass")
```

```python
sample1.correlation("Survived", "Pclass")
```

Из анализа выше видно, что существует достаточно сильная зависимость между классами переменной Pclass и вероятностью выживания: 63% для Pclass = 1, 47% для Pclass = 2, 24% для Pclass = 3
