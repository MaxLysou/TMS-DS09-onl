1. Напишите функцию для разделения строки и преобразования ее в список слов. Пример "Robin Singh" ==> ["Robin", "Singh"]

```python
def split_str(st: str) -> list[str]:
  st = st + ' '
  st_lst = []
  word = ''
  for i in st:
    if i in ' ,;:-.!?': #список разделитей
      if len(word) > 0:
          st_lst.append(word)
      word = ''
    else:
      word += i
  return st_lst

print(f"Введите строку:")
st = str(input())
print(f"Список слов:", split_str(st))
```

 3. Создайте функцию, которая возвращает список целых чисел от n до 1,
 где n>0. Пример: n=5-->[5,4,3,2,1]

```python
def lst(n: int) -> list[int] | str:
  if n > 0:
    lst = list(range(n, 0, -1))
    return lst
  else:
    message = "Ошибка ввода! Длина списка должна быть больше 0!"
    return message

try:
  print(f"Введите длину списка:")
  n = int(input())
  print(f"Список:", lst(n))
except ValueError:
  print('Проверьте введенные данные!')
```

 4. Учитывая год, верните столетие, в котором он находится.
 Пример 1705--> 18 1900--> 19 1601--> 17

```python
def century(year: int) -> int | str:
  if year > 0:
    if year % 100 > 0:
     century = year // 100 + 1
    else:
     century = year // 100
    return century
  else:
    message = "Ошибка ввода! Год не может быть отрицательным!"
    return message

try:
  print(f"Введите год:")
  year = int(input())
  print(year, f"год находится в", century(year), f"столетии.")
except ValueError:
  print('Проверьте введенные данные!')
```

 5. Учитывая случайное неотрицательное число, вам необходимо вернуть
 цифры этого числа в массиве в обратном порядке. 35231 => [1,3,2,5,3]

```python
def reversed_lst(num: int) -> list[int] | str:
    if num >= 0:
      lst = []
      st = str(num)
      for i in range(len(st),0,-1):
        lst.append(int(st[i-1]))
      return lst
    else:
      message = "Ошибка ввода! Введите неотрицательное число!"
      return message

try:
  print(f"Введите неотрицательное число:")
  n = int(input())
  print(f"Полученный список:", reversed_lst(n))
except ValueError:
  print('Проверьте введенные данные!')
```
