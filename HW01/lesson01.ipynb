```python
def table_pivot(param):
    piv=sample.pivot_table(index=param,values='SalePrice',aggfunc={'SalePrice': ['count', 'mean']},dropna=False)
    piv.sort_values('mean', ascending=True, inplace=True)
    piv.insert(1,'percent',piv['count']/len(sample)*100)
    print(piv.round(2))

for column in var_object:
    table_pivot(column)
```

```python
# создаём таблицу
fig, axes = plt.subplots(13, 3, figsize=(15, 50))

# перебираем столбцы и заполняем подграфики
for i, column in enumerate(var_not_object):
    row = i // 3  # вычисляем строку
    col = i % 3   # вычисляем столбец
    sns.histplot(x=column, data=sample, ax=axes[row, col])
    
# Отображаем графики
plt.tight_layout()
plt.show()
plt.close()
```

```python
def func_pivot(param):
    print(param)
    piv=sample.pivot_table(index=param,values='SalePrice',aggfunc={'SalePrice': ['count', 'mean']},dropna=False)
    piv.sort_values('mean', ascending=True, inplace=True)
    piv.insert(1,'percent',piv['count']/len(sample)*100)
    print(piv.round(2))
    
    fig, ax1 = plt.subplots(figsize=(10, 3))
    sns.lineplot(data=piv,x=param,y="mean",ax=ax1,color='blue',label='mean')
    ax2 = ax1.twinx()
    sns.histplot(data=sample,x=param,ax = ax2,color='green',label='count')
    ax1.legend(loc='upper left')
    ax2.legend(loc='upper right')

    plt.show()
    plt.close()



for column in var_object:
    func_pivot(column)
```

```python
# создаём таблицу из графиков
fig, axes = plt.subplots(15, 3, figsize=(15, 50))

# перебираем столбцы и заполняем подграфики
for i, column in enumerate(var_object):
    row = i // 3  # вычисляем строку
    col = i % 3   # вычисляем столбец
    piv=sample.pivot_table(index=column,values='SalePrice',aggfunc={'SalePrice': ['count', 'mean']},dropna=False)
    piv['mean']=piv['mean']/1000
    piv.sort_values('mean', ascending=True, inplace=True)
    piv.insert(1,'percent',piv['count']/len(sample)*100)
    sns.lineplot(data=piv,x=column,y="mean",color='blue',label='mean',ax=axes[row, col])
    sns.histplot(data=sample,x=column,color='green',label='count',ax=axes[row, col])
    
# Отображаем графики
plt.tight_layout()
plt.show()
plt.close()
```
