1. Написать программу для решения уравнений

```python
def equation(a,b,c):
    D = b*b - 4*a*c
    x1 = (-b + D**(0.5))/(2*a)
    x2 = (-b - D**(0.5))/(2*a)
    print(f"x1 =", x1, f",x2 =", x2)

print(f"equation1")
equation(1,-12,20)
print(f"equation2")
equation(1,17,72)
print(f"equation3")
equation(1,-7,-44)
print(f"equation4")
equation(1,9,8)
print(f"equation5")
equation(1,-2,-63)

print(f"equation6")
equation(1,4,-5)
```

2. Дан произвольный список. Представьте его в обратном порядке.

```python
lst_in = [1,1,2,5,8,9,1000]
print(f"input list: ", lst_in)
lst_out1 = []
for i in range(len(lst_in),0,-1):
    lst_out1.append(lst_in[i-1])
print(f"output list: ", lst_out1)

lst_in = ["start",1,2,"middle",8,9,"end"]
print(f"input list: ", lst_in)
lst_out2 = [lst_in[i-1] for i in range(len(lst_in),0,-1)]
print(f"output list: ", lst_out2)

lst_in = [0,1,2,[1,2,3],8,9,"end"]
print(f"input list: ", lst_in)
lst_in.reverse()
print(f"output list: ", lst_in)
```

3. Напишите функцию change(lst), которая принимает список и меняет местами его первый и последний элемент. В исходном списке минимум 2 элемента.

```python
lst_in = [1, 2, 3, 4, 5]
print(f"input list: ", lst_in)

def change(lst):
    if len(lst) < 2:
        print("The string is too short! Minimum size of the string should be 2.")
    else:
        lst.insert(0,lst[len(lst)-1])
        lst.pop(len(lst)-1)
        lst.insert(len(lst),lst[1])
        lst.pop(1)
        print(f"output list: ", lst)
    
change(lst_in)

lst_in = [11, 12, 13, 14, 15]
print(f"input list: ", lst_in)

def change2(lst):
    if len(lst) < 2:
        print("The string is too short! Minimum size of the string should be 2.")
    else:
        lst_out = []
        for i in range(len(lst)):
            if i == 0:
                lst_out.append(lst[len(lst)-1])
            elif i == len(lst)-1:
                lst_out.append(lst[0])
            else:
                lst_out.append(lst[i])
        print(f"output list: ", lst_out)

change2(lst_in)
```

 4. Николай задумался о поиске «бесполезного» числа на основании списка. Суть оного в следующем: он берет произвольный список чисел, находит самое
 большое из них, а затем делит его на длину списка. Студент пока не придумал, где может пригодиться подобное значение, но ищет у вас помощи в реализации
 такой функции useless(s).

```python
lst_in = [1, 2, 5, 8, 13, 100, 15, 9, -1, 0]
print(f"input list: ", lst_in)

def useless(lst):
    useless_number = lst[0]
    for i in lst:
        if useless_number < i:
            useless_number = i
    useless_number = useless_number / len(lst)
    print(f"useless number: ", useless_number)

useless(lst_in)
```

 5. На входе имеем список строк разной длины. Необходимо написать функцию all_eq(lst), которая вернет новый список из строк одинаковой длины. Длину итоговой строки определяем исходя из самой большой из них. Если конкретная строка короче самой длинной, дополнить ее нижними подчеркиваниями с правого края до требуемого количества символов.

```python
lst_in =  ['крот', 'белка', 'выхухоль', "1", ""]
print(f"input list: ", lst_in)

def all_eq(lst):
    max_length = len(lst[0])
    for i in lst:
        if max_length < len(i):
            max_length = len(i)
    lst_out = [i + "_"*(max_length-len(i)) for i in lst]
    print(f"output list: ",lst_out)

all_eq(lst_in)
```
