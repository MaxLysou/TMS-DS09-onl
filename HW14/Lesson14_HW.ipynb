# 1. Комбинаторика

**Задача 1.** В группе 30 студентов. Необходимо выбрать старосту, заместителя старосты и профорга. Сколько существует способов это сделать?

**Решение:**
*   Выбор старосты из 30 студентов => 30 способов
*   **И** выбор зам. старосты из оставшихся (30 - 1) = 29 студентов => 29 способов
*   **И** выбор профорга из оставшихся (30 - 2) = 28 студентов => 28 способов.

По правилу умножения (И): 30 * 29 * 28 способов.

Данная задача так же решается через поиск количества размещений из 30 элементов по 3, то есть выбор 3 неповторяющихся элементов из 30, где важен порядок их размещения.



```python
k = np.array(3)
n = np.array(30)

answer = perm(n, k)
print(f"Ответ:", int(answer), f"способов")
```

**Задача 2.** Два почтальона должны разнести 10 писем по 10 адресам. Сколькими
способами они могут распределить работу?

**Решение:**
*   письмо1 можем быть распределено между 2 почтальонами => 2 способа
*   **И** письмо2 можем быть распределено между 2 почтальонами => 2 способа
*   **И** т.д. 10 раз

По правилу умножения (И): 2 * 2 * ... (10 раз) ...* 2 способов.

Данная задача так же решается через поиск количества размещений из 2 элементов (= почтальонов) по 10 (= писем), то есть размещение 10 писем между 2 почтальонами, где важно принадлежность письма почтальону.

```python
k = np.array(10)
n = np.array(2)

answer = n**k
print(f"Ответ:", int(answer), f"способов")
```

**Задача 3.** В ящике 100 деталей, из них 30 – деталей 1-го сорта, 50 – 2-го, остальные – 3-го. Сколько существует способов извлечения из ящика одной детали 1-го или 2-го сорта?

**Решение:**
*   Одну деталь (1) первого сорта (30) можно извлечь **30 способами**
*   **ИЛИ** Одну деталь (1) второго сорта (50) можно извлечь **50 способами**  

По правилу сложения (ИЛИ): 30 + 50 = 80 способов.

```python
k = np.array(1)
n = np.array(30)
answer1 = comb(n, k, exact=True, repetition=False)

k = np.array(1)
n = np.array(50)
answer2 = comb(n, k, exact=True, repetition=False)

answer = answer1 + answer2
print(f"Ответ:", int(answer), f"способов")
```

**Задача 10.** Сколько существует четырехзначных чисел (возможно, начинающихся с нуля), сумма цифр которых равна 5?

**Решение:**

Искомое число можно представить следующими комбинациями цифр.
1.   0, 0, 0, 5 => n1 перестановок с повторениями (0 повторяется 3 раза, 5 повторяется 1 раз)
2.   0, 0, 1, 4 => n2 перестановок с повторениями (0 повторяется 2 раза, 1 и 4 повторяются 1 раз)
3.   0, 0, 2, 3 => n2 перестановок с повторениями (0 повторяется 2 раза, 2 и 3 повторяются 1 раз)
4.   0, 1, 1, 3 => n2 перестановок с повторениями (1 повторяется 2 раза, 0 и 3 повторяются 1 раз)
5.   0, 1, 2, 2 => n2 перестановок с повторениями (2 повторяется 2 раза, 0 и 1 повторяются 1 раз)
6.   1, 1, 1, 2 => n1 перестановок с повторениями (1 повторяется 3 раза, 2 повторяется 1 раз)

Решением является сумма возможных чисел для каждой из 6 комбинаций, то есть 2 * n1 + 4 * n2.





```python
n = np.array(4)
n1 = np.array(1)
n2 = np.array(3)
answer1 = factorial(n,exact=True) / (factorial(n1,exact=True) * factorial(n2,exact=True))

n1 = np.array(1)
n2 = np.array(1)
n3 = np.array(2)
answer2 = factorial(n,exact=True) / (factorial(n1,exact=True) * factorial(n2,exact=True) * factorial(n3,exact=True))

answer = 2 * answer1 + 4 * answer2
print(f"Ответ:", int(answer), f"чисел")
```

# 2. Классическая вероятностная модель. Геометрическая вероятность

**Задача 1.** В ящике 5 апельсинов и 4 яблока. Наудачу выбираются 3 фрукта. Какова
вероятность, что все три фрукта – апельсины?

**Решение:**

В ящике 9 фруктов: 5 апельсинов и 4 яблока. Вытягиваются 3 фрукта, порядок неважен. Общее число исходов вытянуть 3 фрукта из 9 равно числу сочетаний из 9 элементов по 3.

Положительным исходом считается вытянуть 3 апельсина из имеющихся в корзине 5 апельсинов. Общее число положительных исходов вытянуть 3 апельсина из 5 равно числу сочетаний из 5 элементов по 3.

```python
a = np.array(3)
b = np.array(9)
n = comb(b, a, exact=True, repetition=False)
print(f"Общее число исходов:", int(n))

a = np.array(3)
b = np.array(5)
m = comb(b, a, exact=True, repetition=False)
print(f"Число положительных исходов:", int(m))

answer = m / n
print(f"Ответ. Вероятность наступления события = ", round(answer,6))
```

**Задача 2.** Преподаватель предлагает каждому из трех студентов задумать любое число от 1 до 10. Считая, что выбор каждым из студентов любого числа из заданных равновозможен, найти вероятность того, что у кого-то из них задуманные числа совпадут.

**Решение:**

Преподаватель предлагает каждому из трех студентов задумать любое число от 1 до 10. Общее число исходов равно числу размещений из 10 цифр по 3 студентам с повторениями.

Положительным исходом считается факт того, что у кого-то из них (двух или всех трех) задуманные числа совпадут. Перейдем к обратному событию, то есть у всех студентов различные задуманные цифры. Общее число таких исходов равно числу размещений из 10 цифр по 3 студентам.

```python
a = np.array(3)
b = np.array(10)
n = b**a
print(f"Общее число исходов:", int(n))

a = np.array(3)
b = np.array(10)
m = n - perm(b, a)
print(f"Число положительных исходов:", int(m))

answer = m / n
print(f"Ответ. Вероятность наступления события = ", round(answer,6))
```

**Задача 3.** Найти вероятность того, что в 8-значном числе ровно 4 цифры совпадают, а остальные различны.

**Решение:**

Общее число способов составления восьмизначных чисел равно числу размещений из 10 цифр по 8 с повторениями, так как важен порядок и цифры в числе могут повторяться.

Положительным исходом считается составления числа, состоящего из 4 совпадающих цифр и еще 4 несовпадающих различных цифр, то есть в сумме 5 различных цифр из 10. Имеем два события:
1.   Одна из 10 возможных цифр повторяется 4 раза, то есть занимает 4 места в числе из 8 символов, то есть число сочетаний без повторений из 8 элементов по 4
2.   Оставшиеся 9 цифр занимают 4 места и не повторяются, то есть число размещений без повторений из 9 элементов по 4.

Оба события необходимо перемножить, так как они наступают одновременно, то есть наступает событие 1 И событие 2.


```python
a = np.array(8)
b = np.array(10)
n = b**a
print(f"Общее число исходов:", int(n))

a = np.array(4)
b = np.array(8)
m1 = 10 * comb(b, a, exact=True, repetition=False)

a = np.array(4)
b = np.array(9)
m2 = perm(b, a)

m = m1 * m2
print(f"Число положительных исходов:", int(m))

answer = m / n
print(f"Ответ. Вероятность наступления события = ", round(answer,6))
```

**Задача 4.** Шесть клиентов случайным образом обращаются в 5 фирм. Найти вероятность того, что хотя бы в одну фирму никто не обратится.

**Решение:**

Шесть клиентов случайным образом обращаются в 5 фирм. Общее число исходов равно числу размещений с повторениями из 5 элементов по 6.

Положительным исходом считается факт того, что хотя бы в одну фирму никто не обратится. Рассмотрим обратный исход, что во все фирмы обратился хотя бы один клиент, то есть в одну фирму обратились 2 клиента (например, в первую), а в оставшиеся 4 фирмы по одному клиенту (2,1,1,1,1), приходим к числу  перестановок из 6 элементов с повторениями. И так 5 раз, для каждой из 5 фирм.

```python
a = np.array(6)
b = np.array(5)
n = b**a
print(f"Общее число исходов:", int(n))

a = np.array(6)
b = np.array(2)
m = n - 5 * factorial(a,exact=True) / factorial(b,exact=True)
print(f"Число положительных исходов:", int(m))

answer = m / n
print(f"Ответ. Вероятность наступления события = ", round(answer,6))
```

**Задача 1.** В ящике 10 красных и 5 синих пуговиц. Вынимаются наудачу две пуговицы. Какова вероятность, что пуговицы будут одноцветными?

**Решение:**

В ящике 10 красных и 5 синих пуговиц. Вынимаются наудачу две пуговицы. Общее число исходов равно числу сочетаний из 15 элементов по 2.

Положительным исходом считается факт того, что пуговицы будут одноцветными, то есть ИЛИ 2 красные, ИЛИ 2 синие. В первом случае имеем число сочетаний из 10 элементов по 2, во втором число сочетаний из 5 элементов по 2 и всё это складываем, так как ИЛИ.

```python
a = np.array(2)
b = np.array(15)
n = comb(b, a, exact=True, repetition=False)
print(f"Общее число исходов:", int(n))

a = np.array(2)
b = np.array(10)
m1 = comb(b, a, exact=True, repetition=False)

a = np.array(2)
b = np.array(5)
m2 = comb(b, a, exact=True, repetition=False)

m = m1 + m2
print(f"Число положительных исходов:", int(m))

answer = m / n
print(f"Ответ. Вероятность наступления события = ", round(answer,6))
```

**Задача 3.** В семье – двое детей. Какова вероятность, что старший ребенок – мальчик, если известно, что в семье есть дети обоего пола?

**Решение:**

Общее количество исходов n: (старший ребенок; младший ребенок) = (мальчик; девочка) и (девочка; мальчик) = 2.

Положительный исход m, что старший ребенок мальчик = 1.

Вероятность, что старший ребенок – мальчик = m / n = 1 / 2 = 0.5
