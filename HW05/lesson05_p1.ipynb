```python
lst = [20,15,10000,5,4,8,9,6,3,1,2,7]
print(f"input list:", lst)

def sort_bubble(lst) -> int:
  cnt = 0
  for i in range(len(lst)-1):
    sorted = 0
    for j in range(len(lst)-1-i):
      if lst[j] > lst[j+1]:
        lst[j], lst[j+1] = lst[j+1], lst[j]
        sorted = 1
      cnt += 1
    if sorted == 0:
      break
  return cnt

cnt = sort_bubble(lst)
print(f"output list:", lst)
print(f"count of iterations:",cnt)
```

 2. Напишите функцию search_substr(subst, st), которая принимает 2 строки и определяет, имеется ли подстрока subst в строке st. В случае нахождения
 подстроки, возвращается фраза «Есть контакт!», а иначе «Мимо!». Должно
 быть найдено совпадение независимо от регистра обеих строк

```python
def search_substr(subst: str, st: str) -> int:
  res = "Мимо!"
  for i in range(len(st)):
    if st[i].lower() == subst[0].lower():
      k = 0
      for j in range(len(subst)):
        if len(subst) <= len(st)-i and st[i+j].lower() == subst[j].lower():
          k += 1
      if k == len(subst):
        res = "Есть контакт!"
        break
  return res

print(f"Attempt 1:", search_substr("ello", "Hello, world, hello!"))
print(f"Attempt 2:", search_substr("HE", "Hello, world, hello!"))
print(f"Attempt 3:", search_substr("hello?", "Hello, world, hello!"))
print(f"Attempt 4:", search_substr("o", "Hello, world, hello!"))
print(f"Attempt 5:", search_substr("ooo", "Hello, world, hello!"))
```

3. На основании строки определить 3 наиболее часто встречаемых символа в
 ней. Пробелы нужно игнорировать (не учитывать при подсчете). Для простоты
 подсчета количества вхождений символов удобно использовать Counter из
 модуля collections.

```python
st = "Hello, world! Hello again!"
print(f"input string:", st)

from collections import Counter

st_counter = Counter(st.replace(" ","").lower())
print(st_counter)

print(st_counter.most_common(3))
```

4. Дана строка в виде случайной последовательности чисел от 0 до 9. Требуется
 создать словарь, который в качестве ключей будет принимать данные числа (т.
 е. ключи будут типом int), а в качестве значений– количество этих чисел в
 имеющейся последовательности. Для построения словаря создайте функцию
 count_it(sequence), принимающую строку из цифр. Функция должна
 возвратить словарь из 3-х самых часто встречаемых чисел.

```python
def count_it(sequence: str) -> dict[str, int]:
  sequence_set = set(sequence)
  sequence_dict = dict()
  for i in sequence_set:
    cnt = 0
    for j in sequence:
      if i == j:
        cnt += 1
    sequence_dict.update({i:cnt})

  print(sequence_dict)
  output_dict = dict()
  for k in range(3):
    top1_value = list(sequence_dict.keys())[0]
    top1_count = sequence_dict.get(top1_value)
    for i,j in sequence_dict.items():
      if j > top1_count:
        top1_value = i
        top1_count = j
    output_dict.update({top1_value:top1_count})
    sequence_dict.pop(top1_value)
  return output_dict


count_it("111111111111111111223334444555556666666")
```
