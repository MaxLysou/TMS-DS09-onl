# ParisHousing

```python
#чтение файла и просмотр загруженных данных
ParisHousing_df = pd.read_csv('/content/ParisHousing.csv')
ParisHousing_df.head(5)
```

```python
#исследование данных
ParisHousing_df.info()
```

```python
ParisHousing_df.describe().round(2)
```

```python
ParisHousing_df.isnull().sum()
```

```python
X = ParisHousing_df.drop("price", axis = 1)
y = ParisHousing_df.price
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
print(X_train.shape,X_test.shape,y_train.shape,y_test.shape)
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=3)
X_train_pca = pca.fit_transform(X_train)
X_test_pca = pca.transform(X_test)
```

```python
# Визуализация
plt.figure(figsize=(8,6))
plt.scatter(X_train_pca[:, 0], X_train_pca[:, 1], c=y_train, edgecolor='none', alpha=0.7, s=40,
            cmap=plt.cm.get_cmap('nipy_spectral', 3))
plt.colorbar()
plt.title('PCA of ParisHousing dataset')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.show()
```

```python
# Визуализация
plt.figure(figsize=(8,6))
plt.scatter(X_train_pca[:, 0], X_train_pca[:, 2], c=y_train, edgecolor='none', alpha=0.7, s=40,
            cmap=plt.cm.get_cmap('nipy_spectral', 3))
plt.colorbar()
plt.title('PCA of ParisHousing dataset')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 3')
plt.show()
```

```python
# Визуализация
plt.figure(figsize=(8,6))
plt.scatter(X_train_pca[:, 1], X_train_pca[:, 2], c=y_train, edgecolor='none', alpha=0.7, s=40,
            cmap=plt.cm.get_cmap('nipy_spectral', 3))
plt.colorbar()
plt.title('PCA of ParisHousing dataset')
plt.xlabel('Principal Component 2')
plt.ylabel('Principal Component 3')
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler

# Нормализация данных
scaler_pca = StandardScaler()
X_train_pca_scaled = scaler_pca.fit_transform(X_train_pca)
X_train_pca_scaled

X_test_pca_scaled = scaler_pca.fit_transform(X_test_pca)
X_test_pca_scaled
```

# loan_data

```python
#чтение файла и просмотр загруженных данных
loan_data_df = pd.read_csv('/content/loan_data.csv')
loan_data_df.head(5)
```

```python
#исследование данных
loan_data_df.info()
```

```python
from sklearn.preprocessing import LabelEncoder

# Кодирование категориальных переменных
label_encoder = LabelEncoder()
loan_data_df['person_gender'] = label_encoder.fit_transform(loan_data_df['person_gender'])
loan_data_df['person_education'] = label_encoder.fit_transform(loan_data_df['person_education'])
loan_data_df['person_home_ownership'] = label_encoder.fit_transform(loan_data_df['person_home_ownership'])
loan_data_df['loan_intent'] = label_encoder.fit_transform(loan_data_df['loan_intent'])
loan_data_df['previous_loan_defaults_on_file'] = label_encoder.fit_transform(loan_data_df['previous_loan_defaults_on_file'])
```

```python
#исследование данных
loan_data_df.info()
```

```python
loan_data_df.head(5)
```

```python
loan_data_df.describe().round(2)
```

```python
loan_data_df.isnull().sum()
```

```python
y = loan_data_df.loan_status
X = loan_data_df.drop("loan_status", axis = 1)
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
print(X_train.shape,X_test.shape,y_train.shape,y_test.shape)
```

```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis as LDA

# 2. Применение LDA
lda = LDA(n_components=1)
X_train_lda = lda.fit_transform(X_train, y_train)
X_test_lda = lda.transform(X_test)
```

```python
# Визуализация
plt.figure(figsize=(8,6))
plt.scatter(X_train_lda[:, 0], X_train_lda[:, 0], c=y_train, edgecolor='none', alpha=0.7, s=40,
            cmap=plt.cm.get_cmap('viridis', 2))
plt.colorbar()
plt.title('LDA of loan_data dataset')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.show()
```

```python
# 3. Применение PCA
pca = PCA(n_components=3)
X_train_pca = pca.fit_transform(X_train)
X_test_pca = pca.transform(X_test)
```

```python
# Визуализация
plt.figure(figsize=(8,6))
plt.scatter(X_train_pca[:, 0], X_train_pca[:, 1], c=y_train, edgecolor='none', alpha=0.7, s=40,
            cmap=plt.cm.get_cmap('viridis', 2))
plt.colorbar()
plt.title('PCA of loan_data dataset')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.show()
```

```python
# Визуализация
plt.figure(figsize=(8,6))
plt.scatter(X_train_pca[:, 0], X_train_pca[:, 2], c=y_train, edgecolor='none', alpha=0.7, s=40,
            cmap=plt.cm.get_cmap('viridis', 2))
plt.colorbar()
plt.title('PCA of loan_data dataset')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 3')
plt.show()
```

```python
# Визуализация
plt.figure(figsize=(8,6))
plt.scatter(X_train_pca[:, 1], X_train_pca[:, 2], c=y_train, edgecolor='none', alpha=0.7, s=40,
            cmap=plt.cm.get_cmap('viridis', 2))
plt.colorbar()
plt.title('PCA of loan_data dataset')
plt.xlabel('Principal Component 2')
plt.ylabel('Principal Component 3')
plt.show()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

lr_pca = LogisticRegression()
lr_pca.fit(X_train_pca, y_train)
y_pred_pca = lr_pca.predict(X_test_pca)
print(f"PCA-based Logistic Regression Accuracy: {accuracy_score(y_test, y_pred_pca):.2f}")

report_pca = classification_report(y_test, y_pred_pca)
print(report_pca)
```

```python
lr_lda = LogisticRegression()
lr_lda.fit(X_train_lda, y_train)
y_pred_lda = lr_lda.predict(X_test_lda)
print(f"LDA-based Logistic Regression Accuracy: {accuracy_score(y_test, y_pred_lda):.2f}")

report_lda = classification_report(y_test, y_pred_lda)
print(report_lda)
```

```python
lr = LogisticRegression()
lr.fit(X_train, y_train)
y_pred = lr.predict(X_test)
print(f"Logistic Regression Accuracy: {accuracy_score(y_test, y_pred):.2f}")

report = classification_report(y_test, y_pred)
print(report)
```
