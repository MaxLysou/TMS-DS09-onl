```python
def one_to_two(name_in):
  name_out = [0] * len(name_in)
  for i in range(len(name_in)):
    name_out[i] = [0] * 10

  for i in range(len(name_in)):
    for j in range(10):
      if j == name_in[i]:
        name_out[i][j] = 1
  return name_out
```

```python
# Нормализуем данные
train_images = train_images / 255.0
test_images = test_images / 255.0
```

```python
# Создаем модель
model0 = keras.Sequential([
    keras.layers.Flatten(input_shape=(28, 28)),
    keras.layers.Dense(128, activation='relu'),
    keras.layers.Dense(10)
])

# Компилируем модель
model0.compile(optimizer='adam',
              loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
              metrics=['accuracy'])

# Обучаем модель
model0.fit(train_images, train_labels, epochs=10)

# Оценим точность модели
test_loss, test_acc = model0.evaluate(test_images,  test_labels, verbose=2)
print('\nТестовая точность:', test_acc)
```

```python
#loss = poisson, metrics = huber
#Точноcть модели была 99,54% / 97,49%, стала 0,43% / nan
# Создаем модель
model_boston = keras.Sequential([
    keras.layers.Dense(64, activation='relu', input_dim=13),
    keras.layers.Dense(64, activation='relu'),
    keras.layers.Dense(1, activation='linear')
])

# Компилируем модель
model_boston.compile(optimizer='adam',
              loss=['mse'],
              metrics=['mae'])

# Обучаем модель
model_boston.fit(train_features, train_labels, epochs=20, batch_size = 20)

# Оценим точность модели
test_loss, test_acc = model_boston.evaluate(test_features,  test_labels, verbose=2)
print('\nТестовая точность:', test_acc)
```
